```python
import uproot
import numpy as np
import matplotlib.pyplot as plt

# Path to your final ROOT file
root_file_path = "../preselection/testout_2024H0_new.root"

# Open the ROOT file with Uproot
with uproot.open(root_file_path) as file:
    # Access the TTree named "binnedCounts"
    tree = file["ntuple"]

    # Extract the arrays corresponding to branches in the TTree
    run_numbers = tree["Run"].array()
    lumi_sections = tree["LumiBlock"].array()
    bin_counts = tree["binCounts"].array()

    # Example: Loop through each entry and display the binned counts
    for run, lumi, counts in zip(run_numbers, lumi_sections, bin_counts):
        print(f"Run: {run}, Lumi: {lumi}, Binned Counts: {counts}")
        
        # Optionally, you can plot the binned counts for a specific run/lumi
        plt.plot(np.arange(len(counts)), counts, label=f"Run {run}, Lumi {lumi}")

# Show the plot
plt.xlabel('Mass Bin')
plt.ylabel('Counts')
plt.legend()
plt.show()
```
