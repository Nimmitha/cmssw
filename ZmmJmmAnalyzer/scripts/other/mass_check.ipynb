```python
import uproot as up
import awkward as ak
import matplotlib.pyplot as plt
```

```python
file = 'output.root'
tree = up.open(file)['ntuple']
```

```python
tree.show()
```

```python
branches_to_read = ["B_J1_mass", "B_J1_VtxMass", "B_J1_VtxProb"]
```

```python
arrays = tree.arrays(branches_to_read, library='pd')
```

```python
arrays['diff'] = arrays['B_J1_mass'] - arrays['B_J1_VtxMass']
```
