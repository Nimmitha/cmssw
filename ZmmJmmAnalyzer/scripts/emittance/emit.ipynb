```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV file
df = pd.read_csv("scan_file.csv")  # Replace with your actual filename
df = df[df['when'] != 'Late']
```

```python
# Filter X1 and Y1 separately
x1_data = df[df["type"] == "X1"]
y1_data = df[df["type"] == "Y1"]

# Group by step and calculate std deviation of sep
x1_std = x1_data.groupby("step")["sep"].std()
y1_std = y1_data.groupby("step")["sep"].std()

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(x1_std.index, x1_std.values, marker='o', label='X1 Std Dev')
plt.plot(y1_std.index, y1_std.values, marker='s', linestyle='--', label='Y1 Std Dev')

plt.title("Standard Deviation of Separation vs Step")
plt.xlabel("Step")
plt.ylabel("Standard Deviation of Separation")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
df = df[df['fill'] == 9055]

# Set up the plot
plt.figure(figsize=(12, 6))
plt.title("Step vs Separation for X1 and Y1")
plt.xlabel("Separation")
plt.ylabel("Step")

x1_data = df[df["type"] == "X1"]
y1_data = df[df["type"] == "Y1"]

plt.plot(x1_data["sep"], x1_data["step"], marker='o', label=f"X1")
plt.plot(y1_data["sep"], y1_data["step"], marker='s', linestyle='--', label=f"Y1")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
